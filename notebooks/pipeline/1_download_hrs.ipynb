```python
# Imports

import pandas as pd
from shapely import wkt
import geopandas as gpd

from  hrs_botany.enmap_downloader import EnMAPDownloader
```

```python
plots_path = './data/plots.csv'
# 1. Read the CSV with pandas
plots_df = pd.read_csv(plots_path, index_col=0)

# 2. Convert the WKT‐strings into shapely geometries
plots_df['geometry'] = plots_df['geometry'].apply(wkt.loads)

# 3. Turn it into a GeoDataFrame (set the correct CRS)
plots_gdf = gpd.GeoDataFrame(plots_df, geometry='geometry', crs="EPSG:4326")  
```
